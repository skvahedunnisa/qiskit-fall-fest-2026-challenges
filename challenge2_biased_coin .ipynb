```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1, 1)

print(qc.draw())
```

```python
qc.h(0)

print(qc.draw())
```

```python
qc.t(0)

print(qc.draw())
```

```python
qc.h(0)

print(qc.draw())
```

```python
qc.measure(0, 0)

print(qc.draw())
```

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

job = simulator.run(qc, shots=100)
result = job.result()

counts = result.get_counts()
print("Challenge 2 results:", counts)
```

```python
from IPython.display import display

fig = plot_histogram(counts)
display(fig)
```

## Results and Observations

The circuit uses one qubit and the gates H, T, and H, followed by measurement. We executed the circuit for 100 shots using the Qiskit Aer simulator.

**Observed results:**
- Outcome 0: 88 times (88%)
- Outcome 1: 12 times (12%)

**Theoretical probabilities:**
- Outcome 0: approximately 85.4%
- Outcome 1: approximately 14.6%

The observed results are reasonably close to the theoretical probabilities. Small differences are expected when measuring a quantum circuit a finite number of times.

## Conclusion

The experiment demonstrates how quantum interference can produce unequal measurement probabilities. The first Hadamard gate creates a superposition, the T gate changes the relative phase, and the second Hadamard gate converts that phase difference into a biased measurement distribution.

## Challenge Question

**How can quantum gates change the probability of obtaining 0 or 1 when a qubit is measured?**

Quantum gates change the state of a qubit by changing its probability amplitudes and, in some cases, its relative phase. Gates such as the Hadamard gate create superposition, while phase gates such as T can change the relative phase. When another gate is applied, interference can change the final measurement probabilities. In this experiment, the H–T–H circuit produced different probabilities for 0 and 1, creating a biased quantum coin.
