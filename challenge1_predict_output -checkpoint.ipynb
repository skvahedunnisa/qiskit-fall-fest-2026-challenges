```python
#Circuit A : one X gate
from qiskit import QuantumCircuit
qc_a = QuantumCircuit(1,1)
qc_a.x(0)
qc_a.measure(0,0)
print(qc_a.draw())
```

```python
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import matplotlib.pyplot as plt

simulator = AerSimulator()

job = simulator.run(qc_a, shots=100)
result = job.result()

counts_a = result.get_counts()
print("Circuit A results:", counts_a)

plot_histogram(counts_a)
plt.show()
```

```python
print("Counts:", counts_a)
print("Plotting library:", plt)

fig = plot_histogram(counts_a)
display(fig)
```

```python
# Circuit B: Two X gates
qc_b = QuantumCircuit(1, 1)

qc_b.x(0)
qc_b.x(0)

qc_b.measure(0, 0)

print(qc_b.draw())
```

```python
job_b = simulator.run(qc_b, shots=100)
result_b = job_b.result()

counts_b = result_b.get_counts()
print("Circuit B results:", counts_b)
```

```python
# Circuit C: Apply one Hadamard gate
qc_c = QuantumCircuit(1, 1)

qc_c.h(0)

qc_c.measure(0, 0)

print(qc_c.draw())
```

# Challenge 1: Predict the Output

## Objective
To predict and observe the measurement outcomes of three single-qubit quantum circuits using Qiskit.

## Circuit A: One X Gate
**Prediction:** All 100 shots will produce `1`.

**Actual result:** `{'1': 100}`

**Comparison:** The prediction matched the result. The X gate flipped the initial state from |0⟩ to |1⟩.

## Circuit B: Two X Gates
**Prediction:** All 100 shots will produce `0`.

**Actual result:** `{'0': 100}`

**Comparison:** The prediction matched the result. Applying the X gate twice returned the qubit to its initial state |0⟩.

## Circuit C: One Hadamard Gate
**Prediction:** Approximately 50 zeros and 50 ones.

**Actual result:** `{'0': 54, '1': 46}`

**Comparison:** The result was close to the predicted 50–50 distribution. The Hadamard gate created an equal superposition, so both measurement outcomes had a 50% probability.

## Conclusion
The experiment demonstrated that different quantum gates produce different measurement outcomes. One X gate flipped the qubit, two X gates restored its initial state, and the Hadamard gate produced probabilistic measurement outcomes. The predictions agreed with the observed results.
